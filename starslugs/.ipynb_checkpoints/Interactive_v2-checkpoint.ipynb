## Functions <a name="functions"></a>

```python
def star_info(b):
    with output:
        names = submission_box.value #Star system name, type str
        name_list = names.split(", ")
        system_list = []
        for i in range(len(name_list)):
            system = StarSystem(data, name_list[i])
            system_list.append(system)
        table, fig  = print_info_systems(system_list)
        ax2 = fig.add_subplot(111)
        img = plt.imread('3d_visual_star_systems.png')
        ax2.imshow(img)
        ax2.axis('off')
        ax2.set_position([1,0,1,2.5])
        table.scale(1,2)
        table.auto_set_font_size(False)
        table.set_fontsize(10)
        box = fig.axes[0].get_position()
        fig.axes[0].set_position([box.x0, 0.7, box.width, box.height])
        return table
```

```python
def phys_dist(c):
    with output2:
        names = submit_box.value #Star system name, type str
        name_list = names.split(", ")
        system1 = StarSystem(data, name_list[0])
        system2 = StarSystem(data, name_list[1])
        print("Physical distance between " + name_list[0] + " and " + name_list[1])
        print(round(Distance(system1, system2).physical_dist(),2), " pc")
```
